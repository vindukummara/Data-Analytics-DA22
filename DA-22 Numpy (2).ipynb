## creating  sequences , shuffle, Permutation and linearly seperated data

```python
# shuffle() : used for shuffle the elements
# shuffle() : perform  implace operaction
# shuffle will modify the original array
arr1 = np.arange(10)
print(arr1)

np.random.shuffle(arr1)
arr1
```

```python
# permutation(): shuffle elements but returns new shuffled array
# this don't modify the original array
arr1 = np.arange(10)
print(arr1)
arr2 = np.random.permutation(arr1)
arr2
```

```python
arr1 = np.arange(10)
print(arr1)
arr2 = np.random.permutation(arr1)
print(arr2)
print(arr1)
```

# 4. Array shape, Reshape and transpose

```python
south_famous_dresses = np.array([["Lungi", "Dhoti", "Kurti"],["Saree", "Anarkali", "Half Saree"]])
print(south_famous_dresses)
print(south_famous_dresses.shape)
```

```python
updated_arr = south_famous_dresses.reshape(3,2)
print(updated_arr)
```

```python
updated_arr = south_famous_dresses.reshape(1,6)  # 2D
print(updated_arr)
```

```python
updated_arr = south_famous_dresses.reshape(2,3)
print(updated_arr)
```

```python
updated_arr = south_famous_dresses.reshape(2,1,3)  # 3D      # 2*1*3 = 6  2braces 1row 3columns
print(updated_arr)
```

```python
updated_arr = south_famous_dresses.reshape(1,2,1,3) # 4D         # 1*2*1*3 = 6  2braces 1row 3columns
print(updated_arr)
```

```python
# while reshaping no.of elements must be same
```

```python
# ravel() : used for flatten the array, means convert any array to 1D array
arr4 = south_famous_dresses.ravel()
print(arr4)
```

```python
# transpose : convert Row -> Columns -> Row
print(south_famous_dresses)
print(south_famous_dresses.T)  # T = Transpose
```

## Modifications, insertion & deleition in numpy

```python
import numpy as np
```

```python
south_sweet = np.array([["Motichoor", "Kalakanda", "Jelabi"],
                         ["Bobbatlu", "Kajji Kaya", "Khaja"]])

# Modification
south_sweet[0,0] = 'Payasam'
print(south_sweet)
```

```python
# updated the complete column
south_sweet[1] = ["Gulab jamoon","kesari","Arisallu"]
print(south_sweet)
```

```python
# updated the complete column
# numpy array take fix space to store data
south_sweet[ :, 2] = ['Mysurepak','Doublekamitha']
print(south_sweet)
```

```python
# insert operation : New row
south_sweet = np.insert(south_sweet, 1,['Gajarpak', 'Rasmali', 'Firni'],axis = 0)
print(south_sweet)
```

```python
# insert operation : New column
south_sweet = np.insert(south_sweet, 2, ['Rasbhari', 'Kheer', 'sabdan Kheer'], axis = 1 )
print(south_sweet)
```

```python
# insert operation :multiple columns
south_sweet = np.insert(south_sweet, 2,
                        [['Khajukatli', 'Ghevar', 'Peda'],['Kajukatli','Ghevar', 'Peda']],
                        axis = 1)
print(south_sweet)
                                                
```

```python
# delete operation 
south_sweet = np.delete(south_sweet,0, axis = 0)
print(south_sweet)
```

```python
# delete operation : column
south_sweet = np.delete(south_sweet, 2, axis = 1)
print(south_sweet)
```

```python
# delete operation:column 
south_sweet = np.delete(south_sweet,[1,2], axis = 1)
print(south_sweet)
```
