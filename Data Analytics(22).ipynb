# 1. Data Structure creation (Series (1D array), DataFrame (2D array))  

### Series(1D array)

```python
# Create  Series
ser1 = pd.Series()
print(ser1)               # we use series function
```

```python
# convert  the list to series
list1 = [10, 20, 30, 40, 50]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
```

```python
# convert  the list to series 
# hetroginous elements
# in pandas strings are represented as object
# IMP: In series ,all elements have the same datatype
list1 = [10, 20, 30, 40, 50, "Hello"]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
```

```python
# Convert tuple to Series
tuple1 = (10, 20, 30, 40, 50)
ser2 = pd.Series(tuple1)
print(ser2)
print(type(ser2))
```

```python
tuple1 = (10, 20, 30, 40, 50)
ser2 = pd.Series(tuple1)
print(ser2)
print(type(ser2))
print(ser2[2])
```

```python
tuple1 = (10, 20, 30, 40, 50)
ser2 = pd.Series(tuple1)
print(ser2)
print(type(ser2))
print(ser2[2])
print(ser2[-2])

# In Pandas Series , direct negative indexing not allowed.
```

```python
# Set cannot be converted  to pandas series
# Set i unordered and series is ordered, so no conversion
set1 = {10, 20, 30, 40, 50}
ser2 = pd.Series(set1)
print(ser2)
```

```python
# Convert the numpy array into Series

import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
print(arr1, type(arr1))
```

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
print(arr1, type(arr1))
ser3 = pd.Series(arr1)
print(arr1, type(ser1))
```

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
print(arr1, type(arr1))
ser3 = pd.Series(arr1 , index = ['a' , 'b', 'c', 'd', 'e'])
print(ser3, type(ser3))
print(ser3['a'])
```
