# Bar chart

```python
food_category = ['Biriyani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50,10,20,30]
plt.bar(food_category, like_count, color = ['Pink', 'Green','Black','Orange'])
plt.xlabel('food categories')
plt.ylabel('Like count')
plt.title('food_category like count')
```

```python
food_category = ['Biriyani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50,10,20,30]
plt.figure(figsize = (6,4))
plt.bar(food_category, like_count, color = ['Pink', 'Green','Black','Orange'])

plt.xlabel('food categories')
plt.ylabel('Like count')
plt.title('food_category like count')
```

```python
food_category = ['Biriyani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50,10,20,30]
plt.figure(figsize = (6,4))
plt.bar(food_category, like_count, color = ['Pink', 'Green','Black','Orange'])

# To mention the value places over the bar.
for i, count in enumerate(like_count):
    plt.text(i, count+1, str(count), ha = 'center')

# i : x coordinate.
# count+1 : y-coordinate.
# str(count) : value to print on the chart.
# ha = "center" : Horizontally placing value at the center of the bar. 
plt.xlabel('food categories')
plt.ylabel('Like count')
plt.title('food_category like count')
```

# Histogram

```python
# USE : Histogram works on continuous data. 
# Used to show the frequency of data in a particular range.
# Histogram is popularly used for checking  the shape the data.
```

```python
import numpy as np
data = np.random.randn(1000)  # random normal distributed
plt.figure(figsize = (5,4))
plt.hist(data, bins = 10, edgecolor = 'black')
plt.xlabel('Frequency of data')
plt.ylabel('Continous data')
plt.show()
```

```python
import numpy as np
data = np.random.randn(1000)  # random normal distributed.
plt.figure(figsize = (4,2))
plt.hist(data, bins = 30, edgecolor = 'Red')
plt.show()
```

```python
# Interview Question 
# What is the difference between Bar and Histogram charts?
# Ans:
# 1. Bar chart works on categorical/discrete data while histogram works on continues data.
# 2. Bar chart is used for comparing categories while instagram is used for checking the shape of the data.
```

## Line Chart

```python
# Line Chart: trends over the time 
```

```python
import pandas as pd
import matplotlib.pyplot as plt 

dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')   # ME = Month End
icecream_sales = [90,80,100,120,130,10,20,30,40,70,10,10]
plt.figure(figsize = (4,3))
plt.plot(dates,icecream_sales, marker = '*')

for x,y in zip(dates, icecream_sales):
    plt.text(x,y, str(y))
plt.xticks(dates, dates.strftime("%b"), rotation = 45)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Icecream sales over the months")
plt.grid()
plt.show()
```

```python
dates = pd.date_range(start = "2015-01-01", periods = 12, freq = 'YE')   # YE = Year End
flipkart_sales = [90, 100,10,5,20,40,45,70,90,100,80,40]
amazon_sales = [100,120,130,40,30,20,20,10,5,80,100,150]
Meesho_sales = [10,12,10,40,30,20,120,10,15,18,10,100]
plt.plot(dates,flipkart_sales, marker = 'o',label = "FLIPKART")
plt.plot(dates,amazon_sales, marker = '*', label = "AMAZON")
plt.plot(dates,Meesho_sales, marker = '+', label = "MEESHO")
plt.legend()
```
