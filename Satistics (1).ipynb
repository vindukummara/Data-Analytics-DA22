## Measures of Dispersion (Spread of Data)

```python
# Spreadness of data : How much the data is far away from mean/median.
# MAD, Varience, Standard deviation , Percentile, Quantile, IQR, five number summary.
```

```python
# Mean Absoulte Deviation(MAD)
import numpy as np
Reelcount = [50,100,150,200,500,1000,10,1010]
mean = np.mean(Reelcount)
print(mean)
int_result = []
for i in Reelcount:
    int_result.append(np.abs(i - mean))
print(int_result)


MAD  = (1/len(Reelcount)) * np.sum(int_result)
print(MAD)
```

```python
# Varience
var = (np.var(Reelcount))
print(var)
```

```python
# Stanard deviation
print(np.std(Reelcount))
```

```python
SD = np.sqrt(var)
print(SD)
```

```python
# Percetile
Reelcount = [50,10,15,20,50,70,80,11,5,2]
Q1 = np.percentile(Reelcount,25)
Q2 = np.percentile(Reelcount,50)
Q3 = np.percentile(Reelcount,75)
IQR = Q3 - Q1
LB = Q1 - 1.5 * IQR
UB = Q3 + 1.5 * IQR

print(Q1,Q2,Q3,IQR,UB,LB)
```

```python
* If we repeatedly take random samples from any population and calculation the sample means , the distribution of those sample 
means will tend to follow a normal distribution.
* Even if the original population is not normally distributed -- but each sample size is large (n>=30)
```

### Confidence Interval / Interval Estimation

### 1. Confidence Interval when SD is known (using z table)


```python
import numpy as np
from scipy.stats import norm
# Q1. : find a interval range with 95% confidence

Paneer_pieces = [10, 5, 4, 10, 8, 0, 2]
# CL = 95% , alpha = 0.05,alpha/2 = 0.025, 1-0.025 = 0.975

n = len(Paneer_pieces)
x_bar = np.mean(Paneer_pieces)
pop_sd = np.std(Paneer_pieces, ddof = 0)
sqrt_n = np.sqrt(n)
z_value = norm.ppf(0.975)
inter_value = z_value * (pop_sd / sqrt_n)
CI_Z = (x_bar - inter_value, x_bar + inter_value)
CI_Z
```

```python
Flowers = [11, 1, 3, 5, 0, 6, 15]
# CL = 95% , alpha = 0.05,alpha/2 = 0.025, 1-0.025 = 0.975

n = len(Flowers )
x_bar = np.mean(Flowers )
pop_sd = np.std(Flowers , ddof = 0)
sqrt_n = np.sqrt(n)
z_value = norm.ppf(0.975)
inter_value = z_value * (pop_sd / sqrt_n)
CI_Z = (x_bar - inter_value, x_bar + inter_value)
CI_Z
```

### 2.Confidential Interval when is Unknown (using t table)

```python
#### Confidential Interval when is Unknown using t table
import numpy as np
from scipy.stats import t
# Q1. : find a interval range with 95% confidence

Paneer_pieces = [10, 5, 4, 10, 8, 0, 2]n
# CL = 95% , alpha = 0.05,alpha/2 = 0.025, 1-0.025 = 0.975

n = len(Paneer_pieces)

x_bar = np.mean(Paneer_pieces )
sample_sd = np.std(Paneer_pieces , ddof = 1)
sqrt_n = np.sqrt(n)

t_value = t.ppf(0.975, n-1)
inter_value = t_value * (sample_sd / sqrt_n)

CI_t = (x_bar - inter_value, x_bar + inter_value)
CI_t
```

### two sample t-test

```python
import numpy as np
from scipy.stats import ttest_ind
np.random.seed(42)
male_salary = np.random.randint(60, 100, size = 29)
print(np.mean(male_salary))
female_salary = np.random.randint(200, 400, size = 29)
print(np.mean(female_salary))
t_statistics, p_value = ttest_ind(male_salary, female_salary, alternative = 'less')
print(t_statistics)
alpha = 0.05

if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")
```

### One-Tailed and Two Tailed Tests

1.One-Tailed (Right-Tailed):
* You only care result is greater.
* E.g: Checking if new model's accuracy > 85%. Only care if it got better, not worse.
* One-Tailed test(Left-Tailed):
* you only care if the result is less.
* E.g: Ensuring new updates didn't slow down Wedpage load time. Checking for downgrade only.
* Two-Tailed Test:
* Use this when any change(increase and decrease)is meaningful or risky.
* E.g: Medical testing if a treatment causess any effect. Both improvement or harm are critical.

### T- statistics

```python
# t- satatistics : How much is the difference between sample mean and population mean
# t- statistics: near to 0 : sample mean = population mean
# t-statistics: +ve: sample mean > population mean
# t-statistics : =ve: sample mean < population mean
```

```python
# Null Hypothesis (Claim) : Sample mean is approximatly same as population mean
# Alternate Hypothesis : Sample mean is significanlty different from population mean

import numpy as np
from scipy.stats import ttest_1samp


pop_mean = 87
samples = np.array([87, 88, 86, 89, 90, 84, 84, 88, 89, 90])
print(np.mean(samples))

t_statistics, p_value = ttest_1samp(samples, popmean = pop_mean)
print(t_statistics)
alpha = 0.05

if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")

# p_value is helping to find if reject or fail to reject null hypothesis based upon the alpha value 
```

### 1. one sample Z-test

```python
# Null hypothesis  :sample mean is appproximately equal to population mean 
# Alternate : sample mean is significantly different from the population mean 
from statsmodels.stats.weightstats import ztest

pop_mean = 3.9
pg_food_rating = np.array([
                 3.5,4,2,3,5,
                 2,3.2,3,4,5,4,
                 4,3,3.5,4,6,7,
                 0,5,0,3,5,2,10,
                 1.5,2,3,1,5,4,3,2,
                 1,4,2,1,1,2,3,1])
print(np.mean(pg_food_rating))

# Left tail test
# Alternate hypothesis : sample mean : reject the null hypothesis
# Null hypothesis  : sample mean >= pop mean : Fail to reject the reject the null hypothesis

z_statistics, p_value = ztest(pg_food_rating, value = pop_mean, alternative = 'smaller')
alpha = 0.05

if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")
```

### one propotion ztest

```python
# Null Hypothesis : Restaurant Mehfil claims 90% orders are delivery on time 
# Alternate : claim is wrong 

from statsmodels.stats.proportion import proportions_ztest
claim_propotion = 0.90
sample_orders = 200
on_time = 150
print(on_time/sample_orders)

# Left tail test : sample proportion < claim proportion : Reject null hypothesis
# Null hypothesis  :Fail to reject : sample proportion >= claim proportion

z_statistis , p_value = proportions_ztest(on_time, sample_orders, value = claim_propotion)

alpha = 0.05

if p_value < alpha:
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")
```

```python
# check if the categorical variables are related to each other
# Null hypothesis : Their is no significant relationship between Gender and Buying behaviour 
# Alternate : Their is difference

# Libraries : 
import pandas as pd 
from scipy.stats import chi2_contingency 

# data 
Sale_data = [[10, 50],
             [50,10]]
chi_square, p_value, _, _ = chi2_contingency(Sale_data)

if p_value < 0.05:
    
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")
```

### 2. Goodness of fit:

```python
# Null hypothesis : Predicted class distribution = Expected class distribution
# Alternate : significant difference between predicted and Excepted
from scipy.stats import chisquare 
# Rolling the dice 
Excepted_dice_output = [10,10,10,10,10,10]
# predicted_output = [1,1,1,1,1,55]
predicted_output = [8,9,11,12,9,11]
chi_statistics, p_value = chisquare(f_obs = predicted_output, f_exp = Excepted_dice_output)
if p_value < 0.05:
    
    print("Reject Null hypothesis")
else:
    print("Fail to Reject Null hypothesis")
```

```python
# Null Hypothesis: Their is NO difference between the click rate of version A & B
# Alternate : Their is signicant difference between the click rate of version A & B
from statsmodels.stats.proportion import proportions_ztest
views = [1000, 1000]
clicks = [120, 300]
z_statistics, p_value = proportions_ztest(count = clicks, nobs = views)
alpha = 0.03
if p_value < alpha:
    print("Reject null hypothesis")
else:
    print("Fail to Reject null hypothesis")
```
