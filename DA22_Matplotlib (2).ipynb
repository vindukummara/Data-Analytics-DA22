# Bar chart

```python
food_category = ['Biriyani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50,10,20,30]
plt.bar(food_category, like_count, color = ['Pink', 'Green','Black','Orange'])
plt.xlabel('food categories')
plt.ylabel('Like count')
plt.title('food_category like count')
```

```python
food_category = ['Biriyani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50,10,20,30]
plt.figure(figsize = (6,4))
plt.bar(food_category, like_count, color = ['Pink', 'Green','Black','Orange'])

plt.xlabel('food categories')
plt.ylabel('Like count')
plt.title('food_category like count')
```

```python
food_category = ['Biriyani', 'Idli', 'Dosa', 'Shawarma']
like_count = [50,10,20,30]
plt.figure(figsize = (6,4))
plt.bar(food_category, like_count, color = ['Pink', 'Green','Black','Orange'])

# To mention the value places over the bar.
for i, count in enumerate(like_count):
    plt.text(i, count+1, str(count), ha = 'center')

# i : x coordinate.
# count+1 : y-coordinate.
# str(count) : value to print on the chart.
# ha = "center" : Horizontally placing value at the center of the bar. 
plt.xlabel('food categories')
plt.ylabel('Like count')
plt.title('food_category like count')
```

# Histogram

```python
# USE : Histogram works on continuous data. 
# Used to show the frequency of data in a particular range.
# Histogram is popularly used for checking  the shape the data.
```

```python
import numpy as np
data = np.random.randn(1000)  # random normal distributed
plt.figure(figsize = (5,4))
plt.hist(data, bins = 10, edgecolor = 'black')
plt.xlabel('Frequency of data')
plt.ylabel('Continous data')
plt.show()
```

```python
import numpy as np
data = np.random.randn(1000)  # random normal distributed.
plt.figure(figsize = (4,2))
plt.hist(data, bins = 30, edgecolor = 'Red')
plt.show()
```

## Line Chart

```python
# Line Chart: trends over the time 
```

```python
import pandas as pd
import matplotlib.pyplot as plt 

dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')   # ME = Month End
icecream_sales = [90,80,100,120,130,10,20,30,40,70,10,10]
plt.figure(figsize = (4,3))
plt.plot(dates,icecream_sales, marker = '*')

for x,y in zip(dates, icecream_sales):
    plt.text(x,y, str(y))
plt.xticks(dates, dates.strftime("%b"), rotation = 45)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Icecream sales over the months")
plt.grid()
plt.show()
```

## Pie Chart

```python
# Pie Chart is used on categorical data
# Pie is used to show proportion of the whole
# use pie chart when number of categorices are less 
```

## Multiple sub-charts

```python
# Method-1: using plt.subplot() : create one plot at a time.  good for less  number of subplots.
# Method-2: using plt.subplots() : create all plots at once. good for large number of subsets.
```

## method-1 : using plt.subplot

```python
Dream_breakfast_Thali = ["Idli", "Dosa", "Bonda", "Puri"]
share = [10,40,25,25]
plt.subplot(2, 2, 1)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.subplot(2, 2, 2)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.subplot(2, 2, 3)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.subplot(2, 2, 4)
plt.pie(share, labels = Dream_breakfast_Thali)
```

```python
Dream_breakfast_Thali = ["Idli", "Dosa", "Bonda", "Puri"]
share = [10,40,25,25]
plt.subplot(2, 2, 1)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.subplot(2, 2, 2)
plt.bar(Dream_breakfast_Thali,share)
plt.subplot(2, 2, 3)
plt.bar(Dream_breakfast_Thali,share)
plt.subplot(2, 2, 4)
plt.pie(share, labels = Dream_breakfast_Thali)
```

```python
Dream_breakfast_Thali = ["Idli", "Dosa", "Bonda", "Puri"]
share = [10,40,25,25]
plt.figure(figsize = (8,7))
plt.subplot(2, 3, 1)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.subplot(2, 3, 2)
plt.bar(Dream_breakfast_Thali,share)
plt.subplot(2, 3, 3)
plt.bar(Dream_breakfast_Thali,share)
plt.subplot(2, 3, 4)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.subplot(2, 3, 5)
plt.bar(Dream_breakfast_Thali, share)
plt.subplot(2, 3, 6)
plt.pie(share, labels = Dream_breakfast_Thali)
plt.tight_layout()
plt.show()
```

## method-2 plt.subplots()

```python
Lunch_thali = ["Biryani", "Pulao", "Mutton", "Fish", "Coke"]
share = [50,10,20,15,5]

plt.figure(figsize = (20, 16))
# fig : whole image canvas
# axes : arary of subplots (2 X 3)
fig, axes = plt.subplots(2,3)
for ax in axes.ravel():
    ax.pie(share,labels = Lunch_thali, autopct = "%1.1f%%")
       
plt.tight_layout()              
```

```python
Lunch_thali = ["Biryani", "Pulao", "Mutton", "Fish", "Coke"]
share = [50,10,20,15,5]

plt.figure(figsize = (20, 16))
# fig : whole image canvas
# axes : arary of subplots (2 X 2)
fig, axes = plt.subplots(2,2)
titles = ["thali-1", "thali-2", "thali-3", "thali-4"]
for ax,t in zip(axes.ravel(), titles):     # ravel() convert to 1D
    if t ==  "thali-2":
        ax.bar(Lunch_thali,share)
        ax.set_title(t)
    else:
        ax.pie(share,labels = Lunch_thali, autopct = "%1.1f%%")
        ax.set_title(t)
plt.tight_layout()
```
